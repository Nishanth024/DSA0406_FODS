QUESTION 29

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score
from sklearn.metrics import recall_score, f1_score

data = load_iris()

X = data.data
y = data.target

features = input("Enter feature names separated by commas: ").split(",")
target = input("Enter target variable name: ")

X = data.frame if hasattr(data, "frame") else X

if target != "target":
    print("Using target variable: target")

columns = [name.strip() for name in features]
X = data.data[:, [list(data.feature_names).index(c) for c in columns]]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=1
)

model = DecisionTreeClassifier(random_state=1)
model.fit(X_train, y_train)
pred = model.predict(X_test)

print("Accuracy:", round(accuracy_score(y_test, pred), 2))
print("Precision:", round(precision_score(
    y_test, pred, average="weighted", zero_division=0), 2))
print("Recall:", round(recall_score(
    y_test, pred, average="weighted", zero_division=0), 2))
print("F1-score:", round(f1_score(
    y_test, pred, average="weighted", zero_division=0), 2))
```

QUESTION 32

```python
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split

area = [[1000], [1200], [1500], [1800],
        [2000], [2200], [2500], [2800]]
price = [20, 24, 30, 36, 40, 44, 50, 56]

X_train, X_test, y_train, y_test = train_test_split(
    area, price, test_size=0.25, random_state=1
)

model = LinearRegression()
model.fit(X_train, y_train)
pred = model.predict(X_test)

print("Coefficient:", round(model.coef_[0], 2))
print("Mean Squared Error:", round(mean_squared_error(y_test, pred), 2))
print("R2 Score:", round(r2_score(y_test, pred), 2))

plt.scatter([x[0] for x in area], price)
plt.plot([x[0] for x in area], model.predict(area))
plt.xlabel("House Area")
plt.ylabel("House Price")
plt.title("Area vs House Price")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.DataFrame({
    'Name': ['Arun', 'Bala', 'David', 'John', 'Ravi',
             'Sam', 'Vijay', 'Kumar'],
    'Age': [22, 25, 28, 24, 30, 27, 21, 32],
    'Position': ['Forward', 'Defender', 'Forward', 'Midfielder',
                 'Defender', 'Forward', 'Midfielder', 'Goalkeeper'],
    'Goals': [20, 5, 18, 10, 3, 15, 8, 1],
    'Salary': [50000, 30000, 60000, 40000,
               35000, 55000, 32000, 25000]
})

data.to_csv('players.csv', index=False)

players = pd.read_csv('players.csv')

print("Top 5 players by goals:")
print(players.nlargest(5, 'Goals')[['Name', 'Goals']])

print("\nTop 5 players by salary:")
print(players.nlargest(5, 'Salary')[['Name', 'Salary']])

average_age = players['Age'].mean()
print("\nAverage age:", average_age)

print("\nPlayers above average age:")
print(players[players['Age'] > average_age]['Name'].to_list())

players['Position'].value_counts().plot(kind='bar')
plt.xlabel("Position")
plt.ylabel("Number of Players")
plt.title("Players by Position")
plt.show()
```
